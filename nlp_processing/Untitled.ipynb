In Natural Language Processing, "Love" and "love" are treated as different tokens if the text is case-sensitive. This is because computers consider uppercase and lowercase letters as distinct characters.

For example:

"Love" may appear at the beginning of a sentence or as a proper noun.
"love" is usually used as a common word.

If we do not normalize the case, the model may treat them as two different words, increasing vocabulary size unnecessarily.
To avoid this, we usually convert all text to lowercase, so both become "love", improving consistency and reducing complexity.

What happens if stopwords are not removed?

Stopwords are common words such as "is", "the", "and", "in" that occur frequently but carry less meaningful information.

If stopwords are not removed:

The dataset becomes noisy and large
Important words get less importance
Processing becomes slower
Model performance may decrease because of irrelevant information

However, stopwords are not always useless and must be handled carefully.

Two real-world scenarios where removing stopwords can be harmful
1)Sentiment Analysis

Example:
"I am not happy"
If we remove stopwords like "not", the sentence becomes:
"I am happy"
This completely changes the meaning from negative to positive.

2)Question Answering / Search Systems

Example:
"What is the best course?"
If words like "what", "is", "the" are removed, the sentence structure is lost, making it harder for the system to understand the query properly.

Difference between Stemming and Lemmatization
Stemming
It removes suffixes to get the root form
It is rule-based and fast
Output may not be a real word

Example:

"running" --> "run"
"studies" --> "studi"

Lemmatization
It converts a word to its base (dictionary) form
Uses vocabulary and grammar rules
More accurate but slower

Example:

"running" --> "run"
"better" --> "good"

```python
import re
from collections import Counter

def preprocess_text(text):
    if not text or text.strip() == "":
        return [], ""

    # remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # remove emails
    text = re.sub(r"\S+@\S+", "", text)

    # remove numbers
    text = re.sub(r"\d+", "", text)

    # lowercase
    text = text.lower()

    # remove repeated characters (soooo → soo)
    text = re.sub(r'(.)\1{2,}', r'\1', text)

    # remove special characters & emojis
    text = re.sub(r"[^a-z\s]", "", text)

    # remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    tokens = text.split()

    # remove small words except no/not
    tokens = [w for w in tokens if len(w) > 2 or w in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence


# ✅ Define sentences OUTSIDE the function
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

# ✅ Process each sentence
for s in sentences:
    tokens, clean = preprocess_text(s)

    print("Original:", s)
    print("Tokens:", tokens)
    print("Cleaned:", clean)
    print("-" * 50)
```

```python
for s in sentences:
    tokens, _ = preprocess_text(s)

    if tokens:
        avg_len = sum(len(w) for w in tokens) / len(tokens)
    else:
        avg_len = 0

    print("Sentence:", s)
    print("Total tokens:", len(tokens))
    print("Unique tokens:", len(set(tokens)))
    print("Avg token length:", round(avg_len, 2))
    print("-" * 50)
```

```python
all_tokens = []

for s in sentences:
    tokens, _ = preprocess_text(s)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)

print("Top 10 frequent:", counter.most_common(10))
print("Least frequent:", counter.most_common()[-5:])
```

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
result = full_pipeline(sentences)
print(result)
test_cases = ["", "😂😂😂😂", "123456"]

for t in test_cases:
    print(preprocess_text(t))
```
